```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
RAW_DATA_PATH = "raw_data/loan.csv" 
OUTPUT_DIR = "credit_data"
OUTPUT_FILE = os.path.join(OUTPUT_DIR, "data.csv")

os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
CHUNK_SIZE = 50000
valid_statuses = ['Fully Paid', 'Charged Off', 'Default']
filtered_chunks = []
```

```python
print("Reading and filtering dataset in chunks...")

for i, chunk in enumerate(pd.read_csv(RAW_DATA_PATH, chunksize=CHUNK_SIZE, low_memory=True)):
    # Filter only completed loans
    subset = chunk[chunk['loan_status'].isin(valid_statuses)].copy()
    if not subset.empty:
        filtered_chunks.append(subset)
    print(f"Processed chunk {i + 1} (kept {len(subset):,} rows)")
```

```python
df_filtered = pd.concat(filtered_chunks, ignore_index=True)
del filtered_chunks  # Free memory immediately
print(f"Total completed loans: {df_filtered.shape[0]:,} rows, {df_filtered.shape[1]} columns")
```

```python
df_filtered['target'] = df_filtered['loan_status'].apply(
    lambda x: 1 if x in ['Charged Off', 'Default'] else 0
)

print("\nFull completed dataset target distribution:")
print(df_filtered['target'].value_counts(normalize=True))
```

```python
sample_size = 100000
if len(df_filtered) > sample_size:
    df_sample, _ = train_test_split(
        df_filtered,
        train_size=sample_size,
        stratify=df_filtered['target'],
        random_state=42
    )
else:
    df_sample = df_filtered

del df_filtered
```

```python
print(f"\nFinal sample shape: {df_sample.shape}")
print(f"Sample target distribution:\n{df_sample['target'].value_counts(normalize=True)}")

df_sample.to_csv(OUTPUT_FILE, index=False)
print(f"\nSuccessfully saved 100k sample to: {OUTPUT_FILE}")
```
